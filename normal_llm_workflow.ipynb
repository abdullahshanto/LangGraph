```python
import os
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv(override=True)
```

```python
model = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0)
```

```python
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> dict:
    question = state["question"]
    prompt = f"Answer the following question clearly and concisely: {question}"
    response = model.invoke(prompt)
    return {"answer": response.text}
```

```python
graph = StateGraph(LLMState)

graph.add_node("llm_qa", llm_qa)
graph.add_edge(START, "llm_qa")
graph.add_edge("llm_qa", END)

workflow = graph.compile()
```

```python
initial_state = {"question": "How far is the Moon from the Earth?"}
final_state = workflow.invoke(initial_state)

print(final_state["answer"])
```
